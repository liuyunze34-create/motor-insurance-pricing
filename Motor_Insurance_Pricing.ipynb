# Motor Insurance Pricing

French MTPL pricing project using frequency, severity and pure-premium models. The final test set is held out until model-family selection is complete.

## Setup

The raw data are downloaded from OpenML rather than stored in the repository.

````python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
OUTPUTS = ROOT / "outputs"

summary = pd.read_csv(OUTPUTS / "portfolio_summary.csv")
summary
````

## Data checks

The frequency and severity sources do not reconcile perfectly, so the project reports unmatched records and modeling adjustments explicitly.

````python
pd.read_csv(OUTPUTS / "source_reconciliation.csv")
````

````python
pd.read_csv(OUTPUTS / "modeling_adjustments.csv")
````

## Model comparison

The actuarial benchmarks are a Poisson frequency GLM, Gamma severity GLM and direct Tweedie GLM. A histogram gradient-boosting frequency/severity model is included as a nonlinear challenger. Model-family selection uses repeated validation on the development sample.

````python
stability = pd.read_csv(OUTPUTS / "validation_stability_summary.csv")
stability
````

## Final test results

The selected model is evaluated once on the held-out test set. Gini measures ranking; deviance and calibration are reported separately.

````python
metrics = pd.read_csv(OUTPUTS / "model_metrics.csv")
metrics
````

## Risk deciles

````python
deciles = pd.read_csv(OUTPUTS / "risk_deciles.csv")
deciles[["RiskDecile", "Exposure", "ObservedPurePremium", "PredictedPurePremium", "ActualToPredicted"]]
````
